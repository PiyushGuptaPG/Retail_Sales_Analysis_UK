```python
reg = df.groupby('Region')['Sales_Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
bar = plt.bar(reg.index, reg.values, color='royalblue')
label = (f'{val/1000:.2f}k' for val in reg.values)
plt.bar_label(bar, labels=label, padding=3)
plt.title("Region wise Sales (£) in UK")
plt.xlabel('Regions')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
c_type = df.groupby('Customer_Type')['Sales_Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(4,4))
plt.pie(c_type.values, labels=c_type.index,autopct=lambda p: f'{p:.1f}%',startangle=140, wedgeprops={'edgecolor': 'black'}, colors=plt.cm.Set2.colors)
plt.title("Returning VS New Customers")
plt.tight_layout()
plt.show()
```

```python
df['Month'] = pd.to_datetime(df['Sale_Date']).dt.to_period('M').astype(str)
ms = df[df['Month'] < '2023-12-31']
monthly_sales = ms.groupby('Month')['Sales_Amount'].sum().reset_index()

plt.figure(figsize=(15,5))
sns.lineplot(data=monthly_sales, x='Month', y='Sales_Amount', marker='o', color='royalblue')

for x, y in zip(monthly_sales['Month'],monthly_sales['Sales_Amount']):
    plt.text(x,y +0.1, f'{y/1000:.1f}k', ha='center', va='bottom', fontsize=10)

plt.title('Sales(£) Trend from Jan-2023 to Dec-2023')
plt.xlabel('Months')
plt.ylabel('Sales (£)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
pro = df.groupby('Payment_Method')['Sales_Amount'].sum().sort_values(ascending=False).reset_index()

plt.figure(figsize=(12,5))
cat = sns.barplot(data=pro, x='Payment_Method', y='Sales_Amount', errorbar=None, palette=['midnightblue', 'royalblue', 'skyblue'])
plt.title('Sales (£) By Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Sales (£)')

for container in cat.containers:
    label= [f'{val/1000:.2f}k' for val in container.datavalues]
    cat.bar_label(container, labels=label, padding=3)
```

```python
# ✅ Correct GroupBy
pro = df.groupby('Payment_Method')[['Sales_Amount', 'Profit']].sum().sort_values(by='Sales_Amount',ascending=False).reset_index()

# Melt the dataframe to plot Sales and Profit side by side
pro_melted = pro.melt(id_vars='Payment_Method', value_vars=['Sales_Amount', 'Profit'], 
                      var_name='Metric', value_name='Amount')

plt.figure(figsize=(12,5))
cat = sns.barplot(data=pro_melted, x='Payment_Method', y='Amount', hue='Metric', 
                  palette=['royalblue', 'dodgerblue'], errorbar=None)

plt.title('Sales & Profit by Payment Methods')
plt.xlabel('Payment Method')
plt.ylabel('Amount (£)')

# ✅ Add labels on bars
for container in cat.containers:
    labels = [f'{val/1000:.2f}k' for val in container.datavalues]
    cat.bar_label(container, labels=labels, padding=3)

plt.tight_layout()
plt.show()
```

```python
PCP = df.groupby('Product_Category')['Profit(%)'].median().sort_values(ascending=False).reset_index()

plt.figure(figsize=(8,5))
pro = sns.barplot(data=PCP, x='Product_Category', y='Profit(%)', palette='coolwarm', errorbar=None)

for container in pro.containers:
   label = [f'{val:.2f}%' for val in container.datavalues]
   pro.bar_label(container, labels=label, padding=3)

plt.title("Median Profit(%) earned By Product_Category in 2024")
plt.xlabel("Product_Category")
plt.ylabel('Profit(%)')

plt.tight_layout()
plt.show()
#plt.bar(PCP.index, PCP.values, color='royalblue')
```

```python
df.groupby('Payment_Method').agg(
    total_sales=('Sales_Amount', 'sum'),
    Amount_mean=('Sales_Amount','mean'),
    Amount_median=('Sales_Amount','median'),
    TD = ('Discount','sum'),
    RASP=('Region_and_Sales_Rep', 'count')
)

rr = df.groupby('Sales_Rep').agg(
    total_sales=('Sales_Amount', 'sum'),
    Amount_mean=('Sales_Amount','mean'),
    Amount_median=('Sales_Amount','median'),
    TD = ('Discount','sum'),
    RASP=('Region_and_Sales_Rep', 'count')
)
import matplotlib.pyplot as plt

# Plot total sales by Sales Rep
plt.figure(figsize=(10, 5))
plt.bar(rr.index, rr['Amount_mean'], color='royalblue')

# Add labels & title
plt.xlabel('Sales Rep')
plt.ylabel('Total Sales')
plt.title('Total Sales by Sales Rep')

# Rotate x-labels for better readability
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create subplots: 2 rows, 2 columns
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Sales Rep Performance Metrics', fontsize=16, fontweight='bold')

# Plot 1: Total Sales
axes[0, 0].bar(rr.index, rr['total_sales'], color='royalblue')
axes[0, 0].set_title('Total Sales by Sales Rep')
axes[0, 0].set_ylabel('Total Sales (£)')
axes[0, 0].tick_params(axis='x', rotation=45)

# Plot 2: Mean Sales Amount
axes[0, 1].bar(rr.index, rr['Amount_mean'], color='seagreen')
axes[0, 1].set_title('Average Sales Amount by Sales Rep')
axes[0, 1].set_ylabel('Avg Sales (£)')
axes[0, 1].tick_params(axis='x', rotation=45)

# Plot 3: Median Sales Amount
axes[1, 0].bar(rr.index, rr['Amount_median'], color='orange')
axes[1, 0].set_title('Median Sales Amount by Sales Rep')
axes[1, 0].set_ylabel('Median Sales (£)')
axes[1, 0].tick_params(axis='x', rotation=45)

# Plot 4: Total Discount
axes[1, 1].bar(rr.index, rr['TD'], color='crimson')
axes[1, 1].set_title('Total Discount by Sales Rep')
axes[1, 1].set_ylabel('Total Discount (£)')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout(rect=[0, 0, 1, 0.96])  # Adjust layout to fit title
plt.show()
```
